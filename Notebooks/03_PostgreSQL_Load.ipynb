# Connect Python to PostgreSQL

### Objective

Establish a connection between Python and the Enterprise Banking Data Warehouse using SQLAlchemy.

This connection will be used to load all processed dimension and fact tables into PostgreSQL.

```python
# Import Required Libraries

import pandas as pd
from pathlib import Path
from sqlalchemy import create_engine
```

```python
# Define Processed Folder

project_path = Path.cwd().parent

processed_path = project_path / "Data" / "Processed"
```

```python
# Create PostgreSQL Connection
# Add your PostgreSQL credentials below 

from sqlalchemy import URL, create_engine

connection_url = URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password="YOUR_PASSWORD",
    host="localhost",
    port=5432,
    database="EnterpriseBankingDataPlatform"
)

engine = create_engine(connection_url)
```

```python
# Test Connection

with engine.connect() as conn:
    print("Connected Successfully!")
```

# Load Processed Tables into PostgreSQL

### Objective

Load the processed dimension and fact tables from the local Processed folder into the PostgreSQL warehouse schema.

This step populates the Enterprise Banking Data Warehouse with the transformed data created during the ETL process.

```python
# Create Loading Function

def load_table(file_name, table_name):
    """
    Reads a CSV file and loads it into the PostgreSQL warehouse schema.
    """

    df = pd.read_csv(processed_path / file_name)

    df.to_sql(
        name=table_name,
        con=engine,
        schema="warehouse",
        if_exists="replace",
        index=False
    )

    print(f"✓ Loaded {table_name} ({len(df):,} rows)")
```

```python
# Load Banking Tables

load_table("dim_customer.csv", "dim_customer")
load_table("dim_date.csv", "dim_date")
load_table("dim_merchant.csv", "dim_merchant")
load_table("fact_transactions.csv", "fact_transactions")
```

```python
# Load AML Tables

load_table("dim_account.csv", "dim_account")
load_table("dim_transaction_type.csv", "dim_transaction_type")
load_table("dim_category.csv", "dim_category")
load_table("dim_typology.csv", "dim_typology")
load_table("fact_aml_transactions.csv", "fact_aml_transactions")
```
